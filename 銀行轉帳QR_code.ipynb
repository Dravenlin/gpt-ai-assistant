```python
!pip install qrcode
!pip install pillow

import qrcode
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files
import os

# 生成 QRCode 的函數
def generate_qr():
    # 指定圖片路徑
    image_path = '/content/p2p.png'  # 這裡指定你希望顯示的圖片路徑

    # 確認圖片是否存在
    if not os.path.exists(image_path):
        print(f"找不到圖片文件: {image_path}")
        return

    # 嘗試打開圖片
    try:
        img = Image.open(image_path)
    except Exception as e:
        print(f"無法打開圖片: {e}")
        return

    # 創建繪圖窗口
    fig, ax = plt.subplots(figsize=(6, 8))

    # 顯示圖片
    ax.imshow(img)
    ax.axis('off')  # 不顯示坐標軸
    plt.show()

    # 輸入銀行代碼
    bank_code = input("請輸入銀行代碼 (D5): ")

    # 輸入銀行帳號
    account = input("請輸入銀行帳號 (D6): ").rjust(16, "0")  # 補足為16位，前頭補0

    # 輸入金額
    amount = input("請輸入金額 (D1): ")

    # 驗證金額輸入是否合法
    try:
        amount = f"{int(float(amount) * 100):03d}"
    except ValueError:
        print("請輸入有效的金額")
        return

    # 定義連結
    url = f"TWQRP://這裡隨便要打甚麼都可以/158/02/V1?D1={amount}&D5={bank_code}&D6={account}"

    # 生成 QRCode
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4,
    )
    qr.add_data(url)
    qr.make(fit=True)

    # 自定义 QRCode 的颜色和背景色
    fill_color = "#1E90FF"  # 设置二维码前景色（比如蓝色）
    back_color = "#FFFFFF"  # 设置二维码背景色（白色）

    # 生成带颜色的 QRCode 图片
    qr_img = qr.make_image(fill_color=fill_color, back_color=back_color)

    # 保存 QRCode 图片
    save_path = "/content/qrcode.png"
    qr_img.save(save_path)

    # 顯示 QRCode 圖片
    fig, ax = plt.subplots(figsize=(6, 8))
    ax.imshow(qr_img)
    ax.axis('off')  # 不顯示坐標軸
    plt.show()

    # 提供下載鏈接
    files.download(save_path)

# 呼叫函數生成 QRCode
generate_qr()
```
